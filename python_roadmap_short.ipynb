        * type(): function is basically showing the data type of the numbers.

    * Math:
        * for addition: +
        * for substitution: -
        * for multiplying: *
        * To divide values: /
        * // Floor Divison: it divides two number and rounds down  

        * % remainder: The leftover part after the division and also used to check if a number is even (0) / odd (1)

        +  ** exponentiation: (base ** exponent) It raises a number to the power of another number. to the power function. like, 2{base} to the power 3{exponent} = 8  

    * Rounding:
        * abs(value): returns the absolute (non-negative) value of a number. useful for measuring distance or size.

        * round(): we round the messy numbers, to make results easier to read and work with. it round up to the nearest whole number up or down depending what's closer.

            * round(number, ndigits): rounds the number to the specified number of decimal places

        * Bankers Rounding: ties (like 0.5) go to the nearest even number.
            
        * math.ceil(): rounds a number up to celling.

        * math.floor(): rounds a number down to the floor.

        * math.trunc(): cuts off the decimal part and keeps the whole number (no rounding).

4. > Logic and Operators:
    * Booleans: Build questions(conditions) Python can evaluate as 'yes (True)' or 'no (False)'.

        * Values:
            * True:
            * False:
        * Functions:
            * bool(): 
                * True: if the value is non-empty or non zero
                * False: if the value is empty or zero.
            * any(): Returns True if at least one value is true.
            * all(): Returns True if all values are true.   
            * isinstance(value, type): checks if a value belongs to a certain data type.
        
    * Operators:    
        * Comparison Operators: It's compares two values and return True or False based on the result.
            
            * less then (<)
            * greater then (>)
            * greater then and equal to(>=) 
            * less then and equal to (<=)
            * Equal to (==)
            * Not Equal to (!=)
                * Chain Comparison: check multiple conditions in one line, just like in math. It evaluates it from left to right, checking each condition one by one.    

        * Logical Operators: 
            * and: every conditions must be True then Output will also be True.

            * or: at least one condition must be True then Output will also be True.

            > 'and' operator has higher priority than 'or' operator. 'and' will be executed before 'or'  
            >> for controlling the execution order between 'and','or', we use the parentheses(). 

            * not: It turns True into False and False into True.

        * Membership Operators:
            * in: is a data in a string/group. is 'a' in 'data' it's gonna output True.

            * not in: is a data not in a group/string. 'Rohim' not in ['John', 'Kamal', 'Ali'] output gonna be True.

        * Identify Operators:
            * is: If I store multiple variables with very *simple value* then Python gonna *create same object ID* like one object ID for all the varialbes, all variables going to be *pointing to the same object ID*. *If the ID is same then Output will be True*. If I use 'is not' operator then output will be False.
           
            * is not: If I store multiple variables with very *complex value* then Python gonna *create different object ID* all variables going to be *pointing to the different object ID*. *If the ID isn't same then Output will be False*. If I use 'is' operator then output will be False
